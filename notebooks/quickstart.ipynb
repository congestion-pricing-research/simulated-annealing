# Cordon annealing: quickstart

Synthetic demand example using the installed `cordon_annealing` package. Requires `uv sync --extra dev`.

```
from cordon_annealing import City, AnnealingConfig, optimize
from cordon_annealing.demand import generate_synthetic_demand
from cordon_annealing.initialization import square_by_count
from cordon_annealing.results import save_result
```

```
city = City(8, 8, 1)
city.set_protected_area(city.cell_dict.keys())
demand = generate_synthetic_demand(city, trips_per_cell=12, seed=42)
city.assign_demand_by_cell(demand)
start = square_by_count(city, center=(3.5, 3.5), target_tiles=16)
city.set_enclosed_area(start)
```

```
config = AnnealingConfig(phases=2, steps_unpenalized=10, steps_penalized=10, seed=42)
result = optimize(city, config)
save_result(city, result, config, "../output/quickstart")
```
